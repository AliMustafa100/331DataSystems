Proposition 1

<span style="font-size: 10.5pt; font-family: Arial, sans-serif; font-variant-numeric: normal; font-variant-east-asian: normal; font-variant-alternates: normal; font-variant-position: normal; vertical-align: baseline; white-space-collapse: preserve;">In this query I wanted to see the abbreviation of states located Application.Cities using states from Application.StateProvinces</span>

```sql
SELECT DISTINCT E.CityName , C.StateProvinceCode
From Application.Cities as E
join Application.StateProvinces as C
on E.CityName= C.StateProvinceName
```

<span style="font-size:10.5pt;font-family:Arial,sans-serif;color:#4a4a4a;background-color:#fffffe;font-weight:400;font-style:normal;font-variant:normal;text-decoration:none;vertical-align:baseline;white-space:pre;white-space:pre-wrap;">Proposition 2</span>

<span style="font-family: Arial, sans-serif; font-size: 10.5pt; white-space-collapse: preserve;">In this query we are looking for customers from from both sales.orders and sales.customers and seeing the customers where their creditlimit is not null</span>

```sql
SELECT C.CustomerID, C.CreditLimit, S.OrderID
FROM Sales.Customers AS C
INNER JOIN Sales.Orders AS S
ON C.CustomerID = S.CustomerID
WHERE C.CreditLimit IS NOT NULL
```

<span style="font-size:10.5pt;font-family:Arial,sans-serif;color:#4a4a4a;background-color:#fffffe;font-weight:400;font-style:normal;font-variant:normal;text-decoration:none;vertical-align:baseline;white-space:pre;white-space:pre-wrap;">Proposition 3</span>

<span style="font-family: Arial, sans-serif; font-size: 10.5pt; white-space-collapse: preserve;">We are looking for customerid’s from both sales.orders and sales.customers, we use an inner join to show which customers happen to have 90391 as their zip code. Basically we are checking every customerid from both tables finding&nbsp; a match, and putting it in the table as long as they have their zipcode as 90391</span>

```sql
SELECT C.CustomerID, C.DeliveryPostalCode
FROM Sales.Orders  AS S
INNER JOIN Sales.Customers as C
ON  S.CustomerID = C.CustomerID  
WHERE C.DeliveryPostalCode = 90391
```

<span style="font-family: Arial, sans-serif; font-size: 10.5pt; white-space-collapse: preserve; color: var(--vscode-foreground);">Proposition 4</span>

<span style="font-family: Arial, sans-serif; font-size: 10.5pt; white-space-collapse: preserve; color: var(--vscode-foreground);">Basically the same query but this time we are using left join to show both customerid similarities as well as the leftover from the left table</span>

```sql
SELECT C.CustomerID, S.OrderID
FROM Sales.Customers AS C
LEFT JOIN Sales.Orders AS S
ON C.CustomerID = S.CustomerID;
```

<span style="font-size:10.5pt;font-family:Arial,sans-serif;color:#4a4a4a;background-color:#fffffe;font-weight:400;font-style:normal;font-variant:normal;text-decoration:none;vertical-align:baseline;white-space:pre;white-space:pre-wrap;">Proposition 5</span>

<span style="font-family: Arial, sans-serif; font-size: 10.5pt; white-space-collapse: preserve;">This query is basically checking the people that have tailspintoys in their email or logonname and displaying them</span>

```sql
SELECT A.LogonName, B.BuyingGroupName, A.FullName, A.PersonID
FROM Application.People AS A
JOIN Sales.BuyingGroups AS B
ON A.LogonName LIKE '%tailspintoys%' and B.BuyingGroupName LIKE '%tailspin%'
```

<span style="font-size:10.5pt;font-family:Arial,sans-serif;color:#4a4a4a;background-color:#fffffe;font-weight:400;font-style:normal;font-variant:normal;text-decoration:none;vertical-align:baseline;white-space:pre;white-space:pre-wrap;">Proposition 6</span>

<span style="font-family: Arial, sans-serif; font-size: 10.5pt; white-space-collapse: preserve;">This query shows every customer, and their orders if they have any and if they don’t then lists the customer anyway with no order ID</span>

```sql
SELECT C.CustomerID, C.CustomerName, O.OrderID
FROM Sales.Orders AS O
RIGHT JOIN Sales.Customers AS C
ON O.CustomerID = C.CustomerID;
```

<span style="font-size:10.5pt;font-family:Arial,sans-serif;color:#4a4a4a;background-color:#fffffe;font-weight:400;font-style:normal;font-variant:normal;text-decoration:none;vertical-align:baseline;white-space:pre;white-space:pre-wrap;">Proposition 7</span>

<span style="font-family: Arial, sans-serif; font-size: 10.5pt; white-space-collapse: preserve;">The query&nbsp; produces all possible combinations of buying groups and stock items, regardless of whether the group has ever purchased the item.&nbsp;</span>

```sql
SELECT B.BuyingGroupName, P.StockItemName
FROM Sales.BuyingGroups AS B
CROSS JOIN Warehouse.StockItems AS P
ORDER BY B.BuyingGroupaNme, P.StockItemName;
```

<span style="font-size:10.5pt;font-family:Arial,sans-serif;color:#4a4a4a;background-color:#fffffe;font-weight:400;font-style:normal;font-variant:normal;text-decoration:none;vertical-align:baseline;white-space:pre;white-space:pre-wrap;">Proposition 8</span>

<span style="font-family: Arial, sans-serif; font-size: 10.5pt; white-space-collapse: preserve;">List every stock item; if it has transactions, show each one, otherwise still show the item with blank transaction details.</span>

```sql
SELECT S.StockItemID,S.StockItemName,T.StockItemTransactionID, T.TransactionOccurredWhen,T.Quantity
FROM Warehouse.StockItems AS S
LEFT OUTER JOIN Warehouse.StockItemTransactions AS T
ON S.StockItemID = T.StockItemID
ORDER BY S.StockItemID;
```

<span style="font-size:10.5pt;font-family:Arial,sans-serif;color:#4a4a4a;background-color:#fffffe;font-weight:400;font-style:normal;font-variant:normal;text-decoration:none;vertical-align:baseline;white-space:pre;white-space:pre-wrap;">Proposition 9</span>

<span style="font-family: Arial, sans-serif; font-size: 10.5pt; white-space-collapse: preserve;">This query helps to show each stock item once, along with the total number of transactions it has where 0 means it has none</span>

```sql
SELECT S.StockItemID, S.StockItemName, COUNT(T.StockItemTransactionID) AS TransactionsAmount
FROM Warehouse.StockItems AS S
LEFT JOIN Warehouse.StockItemTransactions AS T
ON S.StockItemID = T.StockItemID
GROUP BY S.StockItemID, S.StockItemName;
```

<span style="font-family: Arial, sans-serif; font-size: 10.5pt; white-space-collapse: preserve;">The query shows each person together with the name of the city they belong to by matching the person’s CityID with the city’s CityID</span>

```sql
SELECT P.PersonID, P.FullName, C.CityName
FROM Application.People AS P
INNER JOIN Application.Cities AS C
ON P.CityID = C.CityID;
```
